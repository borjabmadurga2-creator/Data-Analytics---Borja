```python
#En primer lugar, vamos a modificar los tipos de datos (dtypes) de las columnas, ya que están todas en string


df_validado["salari_mensual"] = df_validado["salari_mensual"].astype(str).str.replace("€", "").str.strip()
df_validado["salari_mensual"] = df_validado["salari_mensual"].str.replace(",",".")
df_validado["salari_mensual"] = pd.to_numeric(df_validado["salari_mensual"], errors="coerce")

columnas_numericas = [
    "dia_naixement", "mes_naixement", "any_naixement", "nombre_fills", "km_anuals", "consum_mitja_l_100km", "temperatura_mitjana_ciutat"
]

for col in columnas_numericas:
  df_validado[col] = pd.to_numeric(df_validado[col], errors="coerce")

df_validado.info()
```

```python
#visto que tenemos varios datos nulos en diferentes columnas, pero que tienen correlaciónentre ellas, vamos a iniciar por juntar las columnas fills, no_fills i nombre_fills.
#si nombre_fills >0 te_fills = true; si té fills = true nombre_fills >=1, si no_fills = True nombre_fills = 0
df_validado.loc[df_validado["nombre_fills"] > 0, "fills"] = "True"
df_validado.loc[df_validado["nombre_fills"] > 0, "no_fills"] = "False"#si numero_hijos es mayor a 0 a fill se le asigna true y false a no_fills

df_validado.loc[df_validado["no_fills"] == "True", "nombre_fills"] = 0
df_validado.loc[df_validado["no_fills"] == "True", "fills"] = "False"
df_validado.loc[df_validado["fills"] == "False", "nombre_fills"] = 0

#si tiene hijos, pero nº hijos es NaN, le asignamos la mediana de hijos

mediana_hijos = df_validado[df_validado["nombre_fills"] > 0 ]["nombre_fills"].median()
df_validado.loc[(df_validado["fills"] == "True") & (df_validado["nombre_fills"].isnull()), "nombre_fills"] = mediana_hijos

df_validado["nombre_fills"] = df_validado["nombre_fills"].fillna(0)

df_validado["fills"] = df_validado["nombre_fills"] > 0
df_validado["no_fills"] = df_validado["nombre_fills"] == 0

#nulos restantes
print(df_validado[["fills","no_fills","nombre_fills"]].isnull().sum())
```

```python
# tabla resumen por género y país de orígen
tabla_pivot = df_validado.pivot_table(
    index="genere",#la columna género será el índice
    columns="pais_origen",#el país de origen serán las columnas
    values="salari_mensual",
    aggfunc="mean",#pido que saque el sou mig de la columna salari_mensual
    margins=True,#activo márgenes para que muestre el salario medio por país
    margins_name="Mitjana_total"
).round(2)
#aquí le doy formato a la tabla, axis=None evalúa toda la tabla conjunta, le doy formato de 2 decimales y le incluyo el símbolo €
tabla_pivot.style.background_gradient(cmap="Blues", axis=None).format("{:.2f} €")

```

```python
#dado que el archivo json no es una tabla, hay que mapear los datos con el archivo enriquecido, creando nuevas columnas
#columna para el grupo profesional
df_enriquecido["incremento_base"] = df_enriquecido["grup_professional"].map(#uso .map() para añadir valores acorde a los datos d euna columna
    politica_salarial["base_increment_by_group"]
)
df_enriquecido
```

```python
#mismo proceso para mapear el incremento por hijo
df_enriquecido["incremento_hijo"] = df_enriquecido["nombre_fills"].apply(#uso apply con una función anónima donde aplico una lógica, si la columan nombre_fills > 0 (té fills == true) aplique la norma has_children. de otra forma no_children
    lambda x: politica_salarial["children_adjustment"]["has_children"] if x > 0
    else politica_salarial ["children_adjustment"]["no_children"]
)
df_enriquecido
```

```python
#mapeo por edad
#calculo la edad de cada uno en base a su fecha de nacimiento
df_enriquecido["data_naixement"] = pd.to_datetime(df_enriquecido["data_naixement"])
df_enriquecido["edad"] = 2026 - df_enriquecido["data_naixement"].dt.year#dt.year extraigo solo el año de la fecha y resto el año actual menos el año de nacimiento

#funcion para asignar el incremento según el rango de edad:
def ajuste_edad(edad):
  if pd.isna(edad):
    return 0.0
  elif 18 <= edad <= 29:
    return politica_salarial["age_adjustment"]["18-29"]
  elif 30 <= edad <= 44:
    return politica_salarial["age_adjustment"]["30-44"]
  elif 45 <= edad <= 59:
    return politica_salarial["age_adjustment"]["45-59"]
  elif edad > 60:
    return politica_salarial["age_adjustment"]["60+"]
  return 0.0

#aplicar columna incremento edad
df_enriquecido["incremento_edad"] = df_enriquecido["edad"].apply(ajuste_edad)
df_enriquecido
```

```python
#añadir el incremento por coste de vida
#en este caso nos vamos a basar en la temperatura de la ciudad, ya que las más frías se asocian a las más caras
#funcion para categorizar el coste de vida acorde a la temperatuda
def categoria_coste_vida(temperatura):
  if pd.isna(temperatura):
    return "medium" #si el campo temperatura es nulo, le asignamos valor neutro "medio"
  elif temperatura < 12:
    return "high"
  elif temperatura <= 16:
    return "medium"
  else:
    return "low"
#creacion de la categoria
df_enriquecido["coste_de_vida"] = df_enriquecido["temperatura_mitjana_ciutat"].apply(categoria_coste_vida)
#mapeo acorde a json
df_enriquecido["inc_cost_vida"] = df_enriquecido["coste_de_vida"].map(
    politica_salarial["cost_of_living_adjustment"]
)
df_enriquecido
```

```python
#ajuste por transporte me baso en la columna km_anuals
#funcion para asignar el ajuste por transporte
def ajuste_transporte(km):
  if pd.isna(km) or km == 0:#si el valor es nulo o es igual a 0 le pongo la categoria low
    return politica_salarial["transport_adjustment"]["low"]
  elif km > 30000:
    return politica_salarial["transport_adjustment"]["high"]
  else:
    return politica_salarial["transport_adjustment"]["medium"]

df_enriquecido["incremento_transporte"] = df_enriquecido["km_anuals"].apply(ajuste_transporte)
df_enriquecido
```

```python
#aplicación de las reglas
#suma de todos los incrementos
df_enriquecido["incremento_total"] = (
    df_enriquecido["incremento_base"] +
    df_enriquecido["incremento_hijo"] +
    df_enriquecido["incremento_edad"] +
    df_enriquecido["incremento_transporte"] +
    df_enriquecido["inc_cost_vida"]
)
#regla incremento máx 15%
max_inc = politica_salarial["rules"]['max_total_increment_percent']
df_enriquecido["inc_total_porcentaje"] = df_enriquecido["incremento_total"].clip(upper=max_inc)

#calcular nuevo salario
df_enriquecido["nou_salari"] = df_enriquecido["salari_mensual"] * (1 + df_enriquecido["inc_total_porcentaje"] / 100)

#aplicar umbral mínimo de salario (min 800)
min_sal = politica_salarial["rules"]["min_salary_threshold"]
df_enriquecido["nou_salari"] = df_enriquecido["nou_salari"].clip(lower=min_sal).round(2)
df_enriquecido

```

La integración de fuentes externas (CSV y JSON) permitió parametrizar la política retributiva de forma dinámica. Ante la presencia de datos parciales o nulos (como fechas de nacimiento omitidas o ciudades que no cruzaron correctamente), se aplicó un criterio conservador y neutro (asignación de incremento 0% o tramo medio). Esto evita sesgos en el cálculo del nou_salari y mantiene la coherencia retributiva sin detener el pipeline de procesamiento.

Por ejemplo, un histograma/boxplot si la variable es numérica, unas barras de los valores más frecuentes si es categórica y unas barras de los años más frecuentes si el dato está en formato fecha.

Nota

Muestra el resultado de la función en alguno de los datasets de ejemplo que contiene el paquete seaborn. Por ejemplo: iris, penguins o titanic.

```python
def exploracion_visual_automatizada(df, output_folder="graficos_exploracion"):

    # Crear la carpeta para dejar las imágenes si no existe
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    #configurar el estilo general
    sns.set_theme(style="whitegrid")

    for col in df.columns:
        plt.figure(figsize=(8, 4))
        plt.title(f"Distribución de la variable: '{col}'", fontsize=12, fontweight="bold")
    #aqui la función decide qué gráfico muestra según el tipo de información que tiene la columna
        #variable tipo fecha
        if pd.api.types.is_datetime64_any_dtype(df[col]):#decide si la columna tiene fecha
            #extracción del año y contar registros por año
            anys = df[col].dt.year.value_counts().head(10)
            sns.barplot(x=anys.index.astype(str), y=anys.values, palette="mako")#grafico de barras y rotación de etiquetas
            plt.xlabel("Any")
            plt.ylabel("Freqüència")
            plt.xticks(rotation=45)

        #variable numérica
        elif pd.api.types.is_numeric_dtype(df[col]):#detección de numeros
            #histograma con intervalo de números con una linea azul que muestre la tendencia
            sns.histplot(df[col], kde=True, color="skyblue")
            plt.xlabel(col)
            plt.ylabel("Freqüència")

        #variable categórica o texto
        else:
            #selecciona 10 valores más frecuentes
            top_valors = df[col].astype(str).value_counts().head(10)
            sns.barplot(
                x=top_valors.values, y=top_valors.index, palette="viridis"
            )
            plt.xlabel("Freqüència")
            plt.ylabel(col)

        plt.tight_layout()#ajuste automático de márgenes

        # Guardar la imagen
        file_name = f"{output_folder}/{col}_grafico.png"
        plt.savefig(file_name, dpi=300)
        print(f"Grafico guardado: {file_name}")

        # Mostrar el grafico por pantalla
        plt.show()



#ejemplo de uso con penguins


# Carregar dataset d'exemple de Seaborn
df_penguins = sns.load_dataset("penguins")



#ejecutar la funcion
exploracion_visual_automatizada(df_penguins)
```

La automatización de la exploración visual mediante detección condicional de tipos de datos (datetime, numeric, object) optimiza la fase inicial de análisis exploratorio (EDA). La función genera un diagnóstico visual completo y persistente en disco (carpeta de imágenes exportadas), permitiendo detectar asimetrías, valores sesgados y distribuciones categóricas sin intervención manual repetitiva.

```python
#carga y revisión de datos
df_distancias = pd.read_excel("/content/matriu_distancies.xlsx",index_col=0)
df_distancias
```

```python
#eliminación de ciudades insulares
ciudades_eliminar = ["Las Palmas de Gran Canaria", "Palma"]
#eliminamos las columnas y filas asociadas a las ciudades a eliminar
df_distancias = df_distancias.drop(
    index = ciudades_eliminar, columns = ciudades_eliminar, errors = "ignore"
)
#muestro dataset sin las dos ciudades
df_distancias
```

```python
#creacion de la funcion de la ruta más cercana
def calcular_ruta_cercana(origen, destino, df):
    if origen not in df.index or destino not in df.index:
        return f"Error: {origen} o {destino} no existen.", 0

    ciudad_actual = origen
    visitadas = [origen]
    distancia_total = 0

    while ciudad_actual != destino:
        # 1. Obtener distancias desde la ciudad actual
        distancias_actuales = df.loc[ciudad_actual]

        # 2. Filtrar solo las ciudades que aún NO se han visitado
        ciudades_no_visitadas = distancias_actuales.drop(labels=visitadas)

        # 3. Buscar la ciudad NO visitada más cercana (idxmin)
        siguiente_ciudad = ciudades_no_visitadas.idxmin()
        distancia = ciudades_no_visitadas[siguiente_ciudad]

        # 4. Actualizar acumuladores
        distancia_total += distancia
        visitadas.append(siguiente_ciudad)
        ciudad_actual = siguiente_ciudad

    return visitadas, distancia_total

#Ejemplo de uso
origen_test = "Málaga"
destino_test = "Barcelona"

ruta, distancia = calcular_ruta_cercana(
    origen_test, destino_test, df_distancias
)

print(f"📍 Ruta seguida: {' ➡️ '.join(ruta)}")
print(f"🚗 Distància total recorreguda: {distancia} km")
```
