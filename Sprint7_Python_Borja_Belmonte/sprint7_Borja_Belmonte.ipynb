Asegurándose de que toda la información introducida sea correcta y entre dentro de los valores permitidos.

A partir de estos datos, el programa deberá calcular una puntuación total aplicando el peso de la experiencia (años * 1.5) y sumando los puntos de catalán, excel y titulación, y a continuación clasificar automáticamente a cada persona según los umbrales establecidos por la empresa:

Prioridad Alta -> puntuación >= 12 Candidato Adecuado -> 8 <= puntuación < 12 No prioritario -> < 8

Cada vez que alguien complete el formulario, se debe mostrar su puntuación individual. Cuando ya se hayan inscrito exactamente diez personas, la herramienta debe dejar de aceptar nuevos registros y generar un informe final en pantalla, ordenado primero según la prioridad y, en caso de empate, por la puntuación obtenida. Este informe debe incluir el nombre, el teléfono y la puntuación de cada candidato para que el departamento de RRHH los pueda contactar rápidamente

```python
# función que obtiene datos de los candidatos

def obtener_datos_candidatos():
  lista_candidatos = []
  contador = 0
  while contador < 10:
    candidato = {
        "nombre_completo" : input("Introduce tu nombre y apellidos: "),
        "telefono" : int(input("Introduce tu número de teléfono: ")),
        "años_experiencia" : int(input("Introduce cuántos años de experiencia laboral tienes: ")),
        "catalan" : int(input("Conocimiento de catalán: \n 0: Sin conocimiento \n 1: Nivel B1 \n 2: Nivel B2 \n 3: Nivel C1 o superior")),
        "excel" : int(input("Dominio de excel: \n 0: Nivel básico \n 1: Uso de funciones \n 2: Tablas dinámicas \n 3: Visual Basic y Macros")),
        "estudios" : int(input("Introduce tu nivel de estudios: \n 0: ESO \n 1: FP \n 2: Universidad \n 3: Master o superior"))
    }
    lista_candidatos.append(candidato)
    contador += 1
  return lista_candidatos
```

```python
# función que valida los datos de los candidatos

def validar_candidatos(lista_candidatos):
  lista_validados = []
  for candidato in lista_candidatos:
    catalan = candidato["catalan"]
    excel = candidato["excel"]
    estudios = candidato["estudios"]
    if (0 <= candidato["catalan"] <= 3) and (0 <= candidato["excel"] <= 3) and (0<= candidato["estudios"] <=3):
      lista_validados.append(candidato)
    else:
      print(f"candidato {candidato["nombre_completo"]} no válido")
  return lista_validados
```

```python
# funcion que calcula la puntuación de los candidatos

def calcular_puntuacion(lista_candidatos):
  for candidato in lista_candidatos:
    puntos_experiencia = candidato["años_experiencia"] * 1.5
    puntuacion_calculada = puntos_experiencia + candidato["catalan"] + candidato["excel"] + candidato["estudios"]
    candidato["puntuacion_total"] = puntuacion_calculada
  return lista_candidatos
```

```python
# función que clasifica a los candidatos

def clasificacion_candidatos(lista_candidatos):
  for candidato in lista_candidatos:
    puntuacion = candidato["puntuacion_total"]
    if puntuacion >=12:
      candidato["clasificacion"] = "prioridad alta"
    elif (8 <= puntuacion < 12):
      candidato["clasificacion"] = "candidato_adecuado"
    else:
      candidato["clasificacion"] = "no prioritario"
  return lista_candidatos
```

```python
# funcion que mostrará el infome de los candidatos

def informe_candidatos(lista_candidatos):
  lista_ordenada = sorted(lista_candidatos, key = lambda x: x["puntuacion_total"], reverse = True)
  for candidato in lista_ordenada:
    print(f"Nombre: {candidato["nombre_completo"]} | Telefono: {candidato["telefono"]} | Puntuacion total: {candidato["puntuacion_total"]} | Clasificacion: {candidato["clasificacion"]}")
```

Dispones de un conjunto de datos con información básica de 10 clientes:

El departamento de CRM ha definido manualmente:

» Fórmula del riesgo final: riesgo_final = factor_codigo x factor_edad

» 2. ¿Qué diferencias hay entre clientes con el mismo comportamiento pero de edades diferentes? (Por ejemplo: ¿un cliente joven que abandona el carrito de compra hace 24 h tiene el mismo riesgo que uno de más edad?)

```python
#funcion que obtiene el factor y el segmento
def obtener_factor_y_segmento(codigo):
  tabla_codigos = {
      0:{"segmento":"Navega pero no añade productos","comportamiento":0.10},
      1:{"segmento":"Añade productos pero no llega al checkout","comportamiento":0.25},
      2:{"segmento":"Llega al checkout pero no hace el pago","comportamiento":0.50},
      3:{"segmento":"Cesta de la compra abandonada hace menos de 24h","comportamiento":0.75},
      4:{"segmento":"Cesta de la compra abandonada hace más de 24h","comportamiento":0.90}
  }
  return tabla_codigos[codigo]
```

```python
#funcion que obtiene el factor edad
def obtener_factor_edad(edad):
  tabla_edad = {
      "18-25":1.20,
      "26-40":1.0,
      "41-60":0.85,
      "60+":0.70
      }
  return tabla_edad[edad]
```

```python
#funcion que calcula el riesgo de cliente
def calcular_riesgo(facotr_segmento,factor_edad):
  return round(facotr_segmento * factor_edad,2)
```

```python
#funcion que procesa a  los clientes
def procesar_clientes(lista_clientes):
  for cliente in lista_clientes:
    datos_comportamiento = obtener_factor_y_segmento(cliente["codi_comportament"])
    cliente["segment"] = datos_comportamiento["segmento"]
    cliente["factor_codi"] = datos_comportamiento["comportamiento"]
    cliente["factor_edad"] = obtener_factor_edad(cliente["grup_edat"])
    cliente["risc_final"] = calcular_riesgo(cliente["factor_codi"], cliente["factor_edad"])
  return lista_clientes
```

```python
#funcion que resuelve las preguntas de marketing digital
def mostrar_respuestas(clientes):
  cliente_max = max(clientes, key=lambda x: x["risc_final"])
  cliente_min = min(clientes, key=lambda x: x["risc_final"])

  print("1. Cliente con mayor riesgo:")
  print(f"   {cliente_max['nom']} (riesgo {cliente_max['risc_final']})")

  print("2. Diferencias entre clientes con mismo código y distinta edad:")
  codigos_distintos = []
  for c in clientes:
    if c["codi_comportament"] not in codigos_distintos:
      codigos_distintos.append(c["codi_comportament"])

  for codigo in codigos_distintos:
    grupo = [c for c in clientes if c["codi_comportament"] == codigo]
    if len(grupo) > 1:
      for c in grupo:
        print(f"   Código {codigo} | {c['nom']} ({c['grup_edat']}) -> riesgo {c['risc_final']}")

  print("3. Clientes que no merecen inversión:")
  descartados = [c for c in clientes if c["codi_comportament"] <= 1]
  for c in descartados:
    print(f"   {c['nom']} (riesgo {c['risc_final']})")

  candidatos_recuperacion = [c for c in clientes if c["codi_comportament"] >= 3]
  cliente_rentable = max(candidatos_recuperacion, key=lambda x: x["risc_final"])
  print("4. Más rentable de recuperar:")
  print(f"   {cliente_rentable['nom']} (código {cliente_rentable['codi_comportament']}, riesgo {cliente_rentable['risc_final']})")

  print(f"El cliente con menor riesgo es {cliente_min['nom']} ({cliente_min['risc_final']})")
```

Trabajas como analista de datos en un centro médico especializado. Varios pacientes han enviado sus síntomas a través de un formulario online, pero los datos llegan en un formato desordenado y poco estructurado, y el equipo médico necesita que los organices para poder priorizar el orden de atención según la gravedad.

El departamento médico te proporciona el diccionario de niveles de riesgo por síntoma, y una lista con información real procedente del sistema.

El objetivo es transformar estos datos en información útil para los médicos. A partir de los registros en bruto, debes:

Convertir cada string en una estructura limpia (diccionario o similar).

Separar y normalizar la lista de síntomas:

Convertir a minúsculas.

Eliminar espacios sobrantes.

Reemplazar espacios intermedios por guiones bajos (por ejemplo, de "fiebre alta" a "fiebre_alta") para que coincidan exactamente con las claves de tu diccionario de riesgos.

Dividir correctamente por comas.

Asignar un riesgo a cada síntoma utilizando el diccionario niveles_riesgo.

Calcular el riesgo total de cada paciente sumando los valores de sus síntomas.

Ordenar a los pacientes de mayor a menor riesgo para determinar la prioridad de atención.

Y un apartado final con los pacientes ordenados por prioridad (más riesgo -> primero).

```python
#funcion que limpia y normaliza los datos de los pacientes
def limpieza_datos(lista_pacientes):
  bloques = lista_pacientes.split("|")
  nom = bloques[0]
  edad = bloques[1]
  sintomas_sucios = bloques[2]

  lista_sintomas = sintomas_sucios.split(",")

  normalizar_sintomas = []
  for sintoma in lista_sintomas:
    sintoma_limpio = sintoma.strip().lower().replace(" ","_")
    normalizar_sintomas.append(sintoma_limpio)

  diccionario_paciente = {
      "nom" : nom,
      "edad" : edad,
      "sintomas" : normalizar_sintomas
  }
  return diccionario_paciente
```

```python
#funcion que calcula el riesgo total de sintomas
def calcular_riesgo_total(sintoma_limpio):
  nivel_riesgo = {
      "fiebre_alta" : 3,
      "dificultad_respirar": 5,
      "dolor_pecho" : 4,
      "tos" : 1,
      "fatiga" : 1,
      "dolor_cabeza" : 1
  }

  sum_total = 0
  for sintoma in sintoma_limpio:
    sum_total = sum_total + nivel_riesgo[sintoma]
  return sum_total
```

```python
#funcion que procesa pacientes
def procesar_pacientes(lista_sucia_pacientes):
  lista_pacientes_procesada = []
  for paciente_sucio in lista_sucia_pacientes:
    paciente_dicc = limpieza_datos(paciente_sucio)
    puntuacion_riesgo = calcular_riesgo_total(paciente_dicc["sintomas"])
    paciente_dicc["riesgo_total"] = puntuacion_riesgo
    lista_pacientes_procesada.append(paciente_dicc)
  return lista_pacientes_procesada
```

```python
#listado de pacientes
def lista_pacientes_sintomas():
  pacients = [
    "Maria|45|tos, Fiebre Alta, dolor_pecho",
    "Luis|33|dolor_cabeza,faTiga",
    "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
    "Jordi|52|fatiga , tos",
    "Anna|29|fiebre alta,dolor_cabeza"
  ]
  return pacients
```

```python
def mostrar_DataFrame(lista_pacientes):
  df = pd.DataFrame(lista_pacientes)
  display(df)
```

```python
#funcion que muestra los pacientes ordenados por gavedad
def main_pacientes():
  pacientes_procesados = procesar_pacientes(lista_pacientes_sintomas())
  pacientes_ordenados = sorted(pacientes_procesados, key=lambda x: x["riesgo_total"], reverse = True)
  mostrar_DataFrame(pacientes_ordenados)
main_pacientes()

```

En este ejercicio se resuelve una necesidad real de un centro médico especializado: priorizar la atención de los pacientes según la gravedad de sus síntomas. Los datos llegan en bruto desde un formulario online, en un formato desordenado (texto plano separado por barras y comas, con mayúsculas y espacios inconsistentes), por lo que el programa primero limpia y normaliza cada registro, separando nombre, edad y síntomas en una estructura de datos clara. A partir de un diccionario de niveles de riesgo proporcionado por el equipo médico, se calcula el riesgo total de cada paciente sumando el peso de sus síntomas, y finalmente se ordena a todos los pacientes de mayor a menor riesgo. Este análisis es clave en un entorno sanitario: permite que el personal médico identifique de un vistazo quién necesita atención inmediata (por ejemplo, un paciente con dificultad respiratoria y fiebre alta) frente a quién puede esperar, optimizando tiempo y recursos limitados y reduciendo el riesgo de que un caso grave pase desapercibido por llegar en un formulario mal estructurado.

```python
#funcion que procesa los clientes
def procesar_cliente(cliente,niveles_seguro):
  nombre = cliente["nombre"]
  edad = cliente["edad"]
  codigo = cliente["codigo"]

  info_seguros = niveles_seguro[codigo]

  categoria = info_seguros["categoria"]
  precio_base = info_seguros["precio_base"]
  fraude = info_seguros["fraude"]

  precio_final = precio_base * (1 + edad /100)

  cliente_procesado = {
      "nombre" : nombre,
      "edad" : edad,
      "categoria" : categoria,
      "precio_base" : precio_base,
      "precio_final" : precio_final,
      "probabilidad_fraude" : fraude
  }
  return cliente_procesado
```

```python
def procesar_grupo_clientes(lista_clientes, niveles_seguros):
  lista_procesada = []

  for cliente in lista_clientes:
    usuario_procesado = procesar_cliente(cliente,niveles_seguros)
    lista_procesada.append(usuario_procesado)

  return lista_procesada
```

Por este motivo, el responsable del departamento te ha pedido desarrollar un generador de contraseñas seguras, totalmente parametrizable y modular.

Requisitos del generador

Debes crear una función principal y todas las subfunciones necesarias para:

Generar contraseñas de longitud variable.

Decidir si deben incluir:

Mayúsculas

Minúsculas

Números

Signos especiales

Asegurarte de que, como mínimo, aparezca un carácter de cada tipo seleccionado.

Nota importante

Explora el funcionamiento del módulo random de la librería numpy.

La función principal del generador debe tener valores por defecto, exactamente los siguientes:

majuscules=True

minuscules=True

numeros=True

signes=False

Estos parámetros son obligatorios y forman parte del diseño del sistema.

```python
def obtener_mayus():
  return string.ascii_uppercase

```

```python
def obtener_minus():
  return string.ascii_lowercase
```

```python
def obtener_numeros():
  return string.digits
```

```python
def obtener_signos():
  return string.punctuation
```

```python
def crear_contrasenya(longitud,majuscules=True,minuscules=True,numeros=True,signes=False):
  tipos_seleccionados = [
      (majuscules, obtener_mayus),
      (minuscules, obtener_minus),
      (numeros, obtener_numeros),
      (signes, obtener_signos)
  ]

  contrasenya = []
  bolsa_total = ""

  for activo, obtener_caracteres in tipos_seleccionados:
    if activo:
      caracteres = obtener_caracteres()
      caracter_obligatorio = np.random.choice(list(caracteres))
      contrasenya.append(str(caracter_obligatorio))
      bolsa_total += caracteres

  caracteres_restantes = longitud - len(contrasenya)
  if caracteres_restantes > 0:
   relleno = np.random.choice(list(bolsa_total), size=caracteres_restantes)
   contrasenya.extend([str(c) for c in relleno])

  mingled_psswd = np.random.permutation(contrasenya)
  psswd = "".join(mingled_psswd)

  return psswd

mi_psswd = crear_contrasenya(12,signes=True)
print(mi_psswd)
```

En este ejercicio se resuelve una necesidad de seguridad informática habitual en cualquier empresa: generar contraseñas robustas y aleatorias para sus usuarios o sistemas internos. El programa permite elegir qué tipos de caracteres debe incluir la contraseña (mayúsculas, minúsculas, números y símbolos especiales), garantizando siempre que aparezca como mínimo un carácter de cada tipo seleccionado, y completando el resto de la longitud pedida con una mezcla totalmente aleatoria de esos mismos tipos. El resultado es una contraseña de la longitud exacta solicitada, sin patrones predecibles, gracias al mezclado final con numpy
