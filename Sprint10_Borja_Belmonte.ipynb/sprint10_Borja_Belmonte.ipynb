```python
#importación librería requests (google colab ya la tiene instalada)
import requests
import json
import pandas as pd
```

```python

#agrego la url a una variable para obtener la lista de publicaciones
url_posts = "https://jsonplaceholder.typicode.com/posts"

lista_publicaciones = requests.get(url_posts)
lista_publicaciones.json()

```

También se podría hacer todo en una sola celda guardando los resultados en una variable e imprimir solo 2 ejemplos de cada

```python
#solución e nuna misma celda

url_publicaciones = "https://jsonplaceholder.typicode.com/posts"

lista_posts = requests.get(url_publicaciones)
publicaciones = lista_posts.json()

url_usuarios = "https://jsonplaceholder.typicode.com/users"
lista_usuarios = requests.get(url_users)
users = lista_usuarios.json()

url_tareas = "https://jsonplaceholder.typicode.com/todos"
lista_tasks = requests.get(url_tareas)
tareas = lista_tasks.json()

print("primeras 2 publicaciones: ")
print(publicaciones[:2])
print("primeros 2 usuarios: ")
print(users[:2])
print("primeras 2 tasks: ")
print(tareas[:2])
```

Muestra por pantalla:

- La cantidad total de cada recurso

```python
#Al tratarse de una lista de Python, se puede usar la función integrada len() para saber el total de cada recurso
publicaciones = lista_publicaciones.json()
usuarios = lista_users.json()
tasks = lista_tareas.json()
print(f"La cantidad de publicaciones total son: {len(publicaciones)}")
print(f"La cantidad de usuarios totales son: {len(usuarios)}")
print(f"la cantidad de tareas totales son: {len(tasks)}")
```

El código 200 muestra que la solicitud fue reconocida por el servidor, aceptada y está siendo procesada

Haz una petición POST para crear una nueva publicación ficticia. Incluye el título, el cuerpo del mensaje y un userId.

- Muestra la respuesta JSON
- Muestra el código de estado


```python
#creación de una nueva publicación ficticia
publi_url = "https://jsonplaceholder.typicode.com/posts/"

datos = {#añado los datos que quiero añadir a la publicación en formato diccionario clave : valor
    "title" : "my first post",
    "body" : "Hello from Python!",
    "userId" : 12
}

response = requests.post(publi_url, json=datos) #paso el resquests POST y serializo los datos con json
print(f"Respuesta JSON: {response.json()}")#muestro por pantalla los resultados en formato json
print(f"Estatus: {response.status_code}")
```

Añado los datos ficticios que quiero introducir mediante un diccionario de Python (clave : valor). Paso el requests POST, la url y los datos creados en formato JSON y muestro por pantalla.
El resultado del código de estado obtenido es 201. La serie de códigos 200 indican que la solicitud fue reconocida por el servidor, aceptada y está siendo procesada.
El código 201 significa "creado" La solicitud se cumplió y el servidor creó un nuevo recurso.

Haz una petiicón PATCH para modificar parcialmente una publicación existente.

- Muestra la respuesta JSON
- Muestra el código de estado

```python
#En este caso, vamos a modificar parcialmente una publicación existente, concretamente, el titulo
modificacion_url = "https://jsonplaceholder.typicode.com/posts/1"#uso la url apuntando al id 1 (primer publicación)

datos_modificados = {
    "title" : "my post has been modified"#le digo qué datos quiero modificar, en este caso el titulo y le digo qué quiero poner en el diccionario python
}

response = requests.patch(modificacion_url, json=datos_modificados)#uso el requests patch para modificar parcialmente la entrada, le paso la url y serializo los datos
print(f"Respuesta JSON: {response.json()}")#muestro los datos
print(f"Estatus: {response.status_code}")
```

Haz una petición DELETE sobre una publicación.

- Muestra la respuesta JSON
- Muestra el código de estado

```python
#en este caso, vamos a eliminar la quinta publicación
eliminacion_url = "https://jsonplaceholder.typicode.com/posts/5"#apunto a la quinta posición

response = requests.delete(eliminacion_url)
print(f"Respuesta JSON: {response.json()}")
print(f"Estatus: {response.status_code}")
```

Vemos que la respuesta es un diccionario vacío, ya que hemos eliminado la publicación número 5. El estatus que muestra es 200 (procesada). En ocasiones también puede sacar estatus 204 "no content", puesto que ha sido eliminada.

Haz una petición GET sencilla:

- Muestra el código de estado de la respuesta

```python
#conocer el código del estado de la consulta. para hacerlo más fácil y ayudarme en el siguiente paso, los guardaré en variables
datos1 = response1.status_code
datos2 = response2.status_code
print(f"El estado de código de la petición 1 es: {datos1}")
print(f"El estado de código de la petición 2 es: {datos2}")
```

Convierte los datos a un DataFrame de pandas:

Utiliza la API de Open BCN para:

- Encontrar un DataSet de tu interés mediante package_search y package_show


```python
#el primer endpoint es con package_search y paso una string para que busque la palabra economia (uso la q con package_search)
url_base = "https://opendata-ajuntament.barcelona.cat/data/api/action/package_search?q=economia"
respuesta = requests.get(url_base)
datos = respuesta.json()
datasets = datos["result"]["results"]
print(len(datasets))
```

En la siguiente celda, hago un bucle anidado for, con un condicional if para que muestre los ids de los datasets que tengan formato csv

```python
for dataset in datasets:
    #print(dataset['title'])
    #print(dataset["id"])
    print(f"{dataset["title"]} - {dataset["id"]}")
    for recurso in dataset["resources"]:
      if recurso["format"] == "CSV":
        print(recurso["id"])
```

De los resultados obtenidos, selecciona uno que tenga recursos disponibles en CSV o JSON


```python
#el segundo endpoint es con packache_show y le paso el id de un dataset con recursos disponibles en csv
url_show = "https://opendata-ajuntament.barcelona.cat/data/api/action/package_show?id=331cba6c-7ca7-49d7-8ec7-19ff39d70aa3"
respuesta_show = requests.get(url_show)
print(json.dumps(respuesta_show.json(), indent=4, ensure_ascii=False))
```

Con el resource_id del recurso seleccionado, realiza una consulta mediante datastore_search para:

- Recuperar, al menos, 100 registros

```python
#utilizo el endpoint datastore_search, con uno de los ids obtenido anteriormente que tengan recursos en csv o json
url_datastore = "https://opendata-ajuntament.barcelona.cat/data/api/action/datastore_search?resource_id=806839e7-6798-48c9-bf04-caf045c959c1"
respuesta_datastore = requests.get(url_datastore)
datos_ds = respuesta_datastore.json()
print(len(datos_ds['result']['records']))
```

Convertir los resultados en DataFrame

```python
#conversion a dataframe
df = pd.json_normalize(datos_ds, record_path=['result','records'])#uso json_normalize para aplanar un json anidado, el record_path es para indicarle la ruta donde debe buscar, en result y dentro, record
df.head()
```
